```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 11)
step = (n >= 0).astype(float)

x1 = (0.5**n) * step
x2 = ((-0.8)**n) * step
x3 = (1.1**n) * step

fig, axes = plt.subplots(1, 3, figsize=(13, 2.6))
for ax, x, title in zip(axes, [x1, x2, x3],
    ["(0.5)^n u(n)", "(-0.8)^n u(n)", "(1.1)^n u(n)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title); ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 11)
delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, title in zip(axes, [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title); ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

```python
n = np.arange(0, 41)
x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

fig, axes = plt.subplots(1, 2, figsize=(13, 2.6))
for ax, x, title in zip(axes, [x1, x2],
    ["cos(2πn/8)", "cos(2πn/7.5)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title); ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

The first one, cos(2 pi n/8), is the one that repeats exactly.
This is because 8 is a whole number. The second one has 7.5, so it cannot repeat exactly
